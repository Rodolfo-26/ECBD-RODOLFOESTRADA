```python
# Porcentaje de valores faltantes por columna
porcentaje_nulos = (df.isnull().sum() / len(df)) * 100
porcentaje_nulos
```

```python
# Detectar registros duplicados
df.duplicated().sum()
```

```python
# Crear copia del dataset
df_limpio = df.copy()
```

```python
# Eliminar registros duplicados
df_limpio = df_limpio.drop_duplicates()
```

```python
df.duplicated().sum()
```

```python
# Rellenar valores nulos en columnas de texto
columnas_texto = ["director", "cast", "country", "date_added", "rating", "duration"]

for columna in columnas_texto:
    if columna in df_limpio.columns:
        df_limpio[columna] = df_limpio[columna].fillna("No disponible")
```

```python
df_limpio.isnull().sum()
```

```python
### Justificación de limpieza
Se eliminaron los registros duplicados para evitar contar el mismo título más de una vez.
Los valores nulos en columnas de texto fueron reemplazados por "No disponible", porque eliminar todas esas filas podría reducir demasiado la cantidad de información útil para el análisis.
```

```python
plt.figure(figsize=(7,4))
sns.countplot(data=df_limpio, x="type")
plt.title("Cantidad de películas y series en Netflix")
plt.xlabel("Tipo de contenido")
plt.ylabel("Cantidad")
plt.show()
```

```python
top_paises = df_limpio[df_limpio["country"] != "No disponible"]["country"].value_counts().head(10)

plt.figure(figsize=(10,5))
sns.barplot(x=top_paises.values, y=top_paises.index)
plt.title("Top 10 países con más contenido en Netflix")
plt.xlabel("Cantidad de títulos")
plt.ylabel("País")
plt.show()
```

```python
df_limpio["date_added"] = df_limpio["date_added"].astype(str)
df_limpio["year_added"] = df_limpio["date_added"].str.extract(r"(\d{4})")
df_limpio["year_added"] = pd.to_numeric(df_limpio["year_added"], errors="coerce")

contenido_por_anio = df_limpio["year_added"].value_counts().sort_index()

plt.figure(figsize=(10,5))
sns.lineplot(x=contenido_por_anio.index, y=contenido_por_anio.values, marker="o")
plt.title("Contenido agregado a Netflix por año")
plt.xlabel("Año")
plt.ylabel("Cantidad de títulos")
plt.show()
```

```python
top_ratings = df_limpio[df_limpio["rating"] != "No disponible"]["rating"].value_counts().head(10)

plt.figure(figsize=(10,5))
sns.barplot(x=top_ratings.index, y=top_ratings.values)
plt.title("Clasificaciones más comunes en Netflix")
plt.xlabel("Clasificación")
plt.ylabel("Cantidad")
plt.xticks(rotation=45)
plt.show()
```

```python
### Conclusiones del análisis DIKW para Netflix

1. Netflix tiene una mayor cantidad de películas que series, lo que muestra que su catálogo se enfoca más en contenido cinematográfico.

2. Estados Unidos aparece como uno de los países con más contenido disponible, reflejando la fuerte presencia de la industria audiovisual estadounidense.

3. El análisis por año permite observar cómo Netflix ha aumentado su catálogo con el paso del tiempo.

4. Las clasificaciones más comunes ayudan a identificar el tipo de público al que se dirige gran parte del contenido.
```
