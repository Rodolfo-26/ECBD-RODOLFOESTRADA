```python
import pandas as pd 
import numpy as np
import kagglehub

from sklearn.model_selection import train_test_split
from sklearn.preprocessing import StandardScaler
from sklearn.neighbors import KNeighborsClassifier
from sklearn.metrics import accuracy_score

import seaborn as sns
import matplotlib.pyplot as plt
```

```python
from sklearn.metrics import (
    accuracy_score,
    confusion_matrix,
    classification_report,
    ConfusionMatrixDisplay
)
```

```python
##Cargamos el dataset iris
from sklearn.datasets import load_iris
iris = load_iris(as_frame=True)
```

```python
print(iris.DESCR)
```

```python
dfiris = iris.frame
```

```python
##Renombrar la columna target
dfiris.rename(columns={"target":"especie"},inplace = True)
```

```python
dfiris["especie"] = dfiris["especie"].map({
    0: "setosa",
    1: "versicolor",
    2: "virginica"
})
```

```python
# variables predictora
X = dfiris.drop (columns=["especie"])
```

```python
print("variables predictoras")
print(X.columns)
```

```python
# Variable objetivo 
y = dfiris["especie"]
```

```python
print("variables objetivo")
print(y.name)
```

## Dividir los datos en prueba y entrenamiento

80% para entrenamiento

20% para prueba

```python
X_train, X_test, y_train, y_test = train_test_split(
    X, y, test_size = 0.20, random_state=42, stratify=y
)
```

```python
print("X_train", X_train.shape)
print("X_test", X_test.shape)
print("y_train", y_train.shape)
print("y_test", y_test.shape)
```

## Escalado de variables

```python
scaler = StandardScaler()
```

```python
# Escalamos las variables 
X_train_scaler = scaler.fit_transform(X_train)
X_test_scaler = scaler.transform(X_test)
```

```python
print("sin escalar")
print(X_train.head())
print("Escalado")
print(X_train_scaler[:5])
```

## Entrenamiento del modelo KNN
En esta prueba utilizaremos K = 3

```python
modelo = KNeighborsClassifier(n_neighbors=3)
```

```python
modelo.fit(
    X_train_scaler, y_train
)
```

##Realizando predicciones con las variables de prueba

```python
prediccion = modelo.predict(X_test_scaler)
```

```python
print(prediccion[:10])
```

```python
#Comparacion entre valores reales y valores predecidos
comparacion = pd.DataFrame({
    "especie real":y_test.values,
    "especie predicha": prediccion
})
```

```python
(comparacion.head(30))
```

## Evaluacion del modelo

```python
accuracy = accuracy_score(y_test, prediccion)
print(f"accuracy/exactitud: {accuracy: .4f}")
```

## Matriz de confusion

```python
cm = confusion_matrix(y_test, prediccion)
```

```python
cm_df = pd.DataFrame(
    cm,
    index = modelo.classes_,
    columns= modelo.classes_
)
cm_df
```

## classification_report
Proporciona metricas que permiten evaluar el desempeño de cada clase: Precision, recall, f1-score, support

```python
print(classification_report(y_test, prediccion))
```
